```python
%%bash
set -e

cd /content
git clone https://github.com/flutter/flutter.git -b stable

echo 'export PATH="$PATH:/content/flutter/bin"' >> ~/.bashrc
export PATH="$PATH:/content/flutter/bin"

flutter --version
```

```python
%%bash
set -e

mkdir -p /content/android-sdk/cmdline-tools
cd /content/android-sdk/cmdline-tools

wget https://dl.google.com/android/repository/commandlinetools-linux-11076708_latest.zip -O tools.zip
unzip tools.zip
mv cmdline-tools latest

export ANDROID_HOME=/content/android-sdk
export PATH=$PATH:$ANDROID_HOME/cmdline-tools/latest/bin
export PATH=$PATH:$ANDROID_HOME/platform-tools

yes | sdkmanager --licenses
```

```python
%%bash
export ANDROID_HOME=/content/android-sdk
export PATH=$PATH:$ANDROID_HOME/cmdline-tools/latest/bin
export PATH=$PATH:$ANDROID_HOME/platform-tools

sdkmanager \
  "platform-tools" \
  "platforms;android-36" \
  "build-tools;28.0.3"
```

```python
%%bash
export PATH="$PATH:/content/flutter/bin"
export ANDROID_HOME=/content/android-sdk

flutter config --android-sdk $ANDROID_HOME
flutter doctor
```
